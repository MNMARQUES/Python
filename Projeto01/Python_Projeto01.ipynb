```python
# Relação dos nomes
nomes = [
    'Maria', 'Julieta', 'Fernando', 'Cristiano',
    'Julieta', 'Maria', 'Fernando', 'Cláudio'
]

# Dicionário que armazenará a quantidade de letras
qtd_letras = {}

for nome in nomes:
    qtd_letras[nome] = len(nome)

print(qtd_letras)
```

```python
def area(r, pi=3.14):
    return pi * (r ** 2)

# Utilização com o valor padrão de pi
print(area(8))

# Utilização com um valor específico de pi
print(area(8, 3.141592))
```

```python
area = lambda r, pi=3.14: pi * (r ** 2)

# Utilização com o valor padrão
print(area(8))

# Utilização com um valor específico de pi
print(area(8, 3.141592))
```

```python

# Encontra e retorna o maior número ímpar da lista
def maior_impar(lista):
    impares = [numero for numero in lista if numero % 2 != 0]
    return max(impares)


# Encontra e retorna o menor número ímpar da lista
def menor_impar(lista):
    impares = [numero for numero in lista if numero % 2 != 0]
    return min(impares)


# Encontra e retorna o maior e o menor número ímpar
def maior_e_menor_impar(lista):
    return maior_impar(lista), menor_impar(lista)
```

```python
numeros = [10, 3, 7, 2, 15, 8, 21, 4]

print(maior_impar(numeros))
print(menor_impar(numeros))
print(maior_e_menor_impar(numeros))
```

```python
cardiologista = {'terca', 'quarta'}
ortopedista = {'terca', 'quinta'}
dermatologista = {'segunda', 'quarta', 'sexta'}
neurologista = {'terca', 'quinta', 'sexta'}
psiquiatra = {'segunda', 'quarta', 'sexta'}
```

```python

# Calcula os dias possíveis para dois médicos
def disp_dois_especialistas(medico01, medico02):
    return medico01 & medico02


# Calcula os dias possíveis para três médicos
def disp_tres_especialistas(medico01, medico02, medico03):
    return medico01 & medico02 & medico03
```

```python
# Ortopedista e neurologista
print(disp_dois_especialistas(ortopedista, neurologista))

# Dermatologista, neurologista e psiquiatra
print(disp_tres_especialistas(
    dermatologista,
    neurologista,
    psiquiatra
))
```
